# EEG BCI Analysis & Classification
## BioAmp EXG Pill + ADS1115 Dataset

This notebook analyzes EEG recordings from the motor imagery BCI system.

**Pipeline**:
1. Load and explore data
2. Signal quality assessment
3. Frequency domain analysis
4. Feature extraction (time + frequency)
5. Machine learning classification
6. Deep learning (CNN)
7. Results visualization

```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import signal
from scipy.stats import kurtosis, skew
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import warnings
warnings.filterwarnings('ignore')

# Plot settings
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
%matplotlib inline

print("✓ Libraries loaded")
```

## 1. Load Data

```python
# Load CSV file
filename = 'eeg_precision_bci.csv'

df = pd.read_csv(filename)
print(f"Dataset shape: {df.shape}")
print(f"Recording duration: {df['Timestamp_Ms'].max() / 1000:.1f} seconds")
print(f"\nSample rate: ~{len(df) / (df['Timestamp_Ms'].max() / 1000):.1f} Hz")
print(f"\nClass distribution:")
print(df['Intent_Label'].value_counts())

df.head()
```

```python
# Extract signal and labels
Fs = 256  # Sampling rate (Hz)
signal_raw = df['Signal_Raw'].values
labels = df['Label_Class'].values
time_s = df['Timestamp_Ms'].values / 1000  # Convert to seconds

# Remove baseline offset
signal_centered = signal_raw - 500.0

print(f"Signal range: [{signal_centered.min():.2f}, {signal_centered.max():.2f}]")
print(f"Signal mean: {signal_centered.mean():.2f}")
print(f"Signal std: {signal_centered.std():.2f}")
```

## 2. Signal Quality Assessment

```python
# Plot raw signal with class labels
fig, axes = plt.subplots(2, 1, figsize=(16, 8), sharex=True)

# Signal plot
axes[0].plot(time_s, signal_centered, 'b-', lw=0.5, alpha=0.7)
axes[0].set_ylabel('EEG Signal (a.u.)', fontsize=12)
axes[0].set_title('Raw EEG Recording', fontsize=14, fontweight='bold')
axes[0].grid(True, alpha=0.3)

# Class labels as colored background
class_colors = {0: 'lightgray', 1: 'lightgreen', 2: 'lightyellow'}
class_names = {0: 'REST', 1: 'CLICK', 2: 'SQUEEZE'}

current_class = labels[0]
start_idx = 0

for i in range(1, len(labels)):
    if labels[i] != current_class:
        # Plot previous segment
        color = class_colors.get(current_class, 'white')
        axes[1].axvspan(time_s[start_idx], time_s[i-1], 
                        color=color, alpha=0.5)
        start_idx = i
        current_class = labels[i]

# Plot last segment
axes[1].axvspan(time_s[start_idx], time_s[-1], 
                color=class_colors.get(current_class, 'white'), alpha=0.5)

axes[1].set_ylabel('Task Label', fontsize=12)
axes[1].set_xlabel('Time (seconds)', fontsize=12)
axes[1].set_title('Task Segmentation', fontsize=14, fontweight='bold')
axes[1].set_ylim(-0.5, 2.5)
axes[1].set_yticks([0, 1, 2])
axes[1].set_yticklabels(['REST', 'CLICK', 'SQUEEZE'])
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"Total trials: {len(np.where(np.diff(labels) != 0)[0]) + 1}")
```

## 3. Frequency Domain Analysis

```python
# Power Spectral Density (Welch's method)
freqs, psd = signal.welch(signal_centered, Fs, nperseg=Fs*2)

# Plot PSD
fig, ax = plt.subplots(figsize=(14, 6))

ax.semilogy(freqs, psd, 'b-', lw=1.5)
ax.set_xlabel('Frequency (Hz)', fontsize=12)
ax.set_ylabel('Power Spectral Density', fontsize=12)
ax.set_title('EEG Power Spectrum', fontsize=14, fontweight='bold')
ax.set_xlim(0, 50)
ax.grid(True, alpha=0.3)

# Highlight EEG bands
bands = {
    'Delta (0.5-4 Hz)': (0.5, 4, 'purple'),
    'Theta (4-8 Hz)': (4, 8, 'blue'),
    'Alpha (8-13 Hz)': (8, 13, 'green'),
    'Beta (13-30 Hz)': (13, 30, 'orange'),
    'Gamma (30-50 Hz)': (30, 50, 'red')
}

for name, (low, high, color) in bands.items():
    ax.axvspan(low, high, alpha=0.15, color=color, label=name)

ax.legend(loc='upper right', fontsize=10)
plt.tight_layout()
plt.show()

# Calculate band powers
print("\nEEG Band Powers:")
for name, (low, high, _) in bands.items():
    idx = np.logical_and(freqs >= low, freqs <= high)
    band_power = np.trapz(psd[idx], freqs[idx])
    print(f"{name:20s}: {band_power:.2e}")
```

```python
# Spectrogram (time-frequency representation)
f, t, Sxx = signal.spectrogram(signal_centered, Fs, nperseg=Fs)

fig, ax = plt.subplots(figsize=(16, 6))
im = ax.pcolormesh(t, f, 10 * np.log10(Sxx), shading='gouraud', 
                   cmap='viridis', vmax=-20)
ax.set_ylabel('Frequency (Hz)', fontsize=12)
ax.set_xlabel('Time (seconds)', fontsize=12)
ax.set_title('EEG Spectrogram (Time-Frequency)', fontsize=14, fontweight='bold')
ax.set_ylim(0, 40)
cbar = fig.colorbar(im, ax=ax)
cbar.set_label('Power (dB)', fontsize=11)
plt.tight_layout()
plt.show()
```

## 4. Feature Extraction

Extract features from **non-overlapping 1-second windows**

```python
def extract_band_power(signal_segment, fs, freq_range):
    """Extract power in a specific frequency band."""
    # Bandpass filter
    sos = signal.butter(4, freq_range, btype='band', fs=fs, output='sos')
    filtered = signal.sosfilt(sos, signal_segment)
    # RMS power
    return np.sqrt(np.mean(filtered**2))

def extract_features(signal_segment, fs=256):
    """Extract comprehensive feature set from signal segment."""
    features = []
    
    # Time-domain features
    features.append(np.mean(signal_segment))          # Mean
    features.append(np.std(signal_segment))           # Standard deviation
    features.append(np.var(signal_segment))           # Variance
    features.append(kurtosis(signal_segment))         # Kurtosis
    features.append(skew(signal_segment))             # Skewness
    features.append(np.max(signal_segment))           # Maximum
    features.append(np.min(signal_segment))           # Minimum
    features.append(np.ptp(signal_segment))           # Peak-to-peak
    
    # Zero crossing rate
    zcr = np.sum(np.diff(np.sign(signal_segment)) != 0) / len(signal_segment)
    features.append(zcr)
    
    # Frequency-domain features (band powers)
    delta_power = extract_band_power(signal_segment, fs, [0.5, 4])
    theta_power = extract_band_power(signal_segment, fs, [4, 8])
    alpha_power = extract_band_power(signal_segment, fs, [8, 13])
    beta_power = extract_band_power(signal_segment, fs, [13, 30])
    gamma_power = extract_band_power(signal_segment, fs, [30, 45])
    
    features.extend([delta_power, theta_power, alpha_power, beta_power, gamma_power])
    
    # Band power ratios (cognitive markers)
    features.append(alpha_power / beta_power if beta_power > 0 else 0)   # Alpha/Beta
    features.append(theta_power / alpha_power if alpha_power > 0 else 0) # Theta/Alpha
    features.append((theta_power + alpha_power) / beta_power if beta_power > 0 else 0)
    
    return np.array(features)

# Segment data into 1-second windows
window_size = Fs  # 256 samples = 1 second
n_windows = len(signal_centered) // window_size

X_features = []
y_labels = []

for i in range(n_windows):
    start = i * window_size
    end = start + window_size
    
    segment = signal_centered[start:end]
    label = labels[start]  # Label at start of window
    
    features = extract_features(segment, Fs)
    X_features.append(features)
    y_labels.append(label)

X = np.array(X_features)
y = np.array(y_labels)

print(f"Feature matrix shape: {X.shape}")
print(f"Number of features per window: {X.shape[1]}")
print(f"Number of windows: {len(y)}")
print(f"\nClass distribution in windows:")
unique, counts = np.unique(y, return_counts=True)
for cls, count in zip(unique, counts):
    print(f"  Class {cls}: {count} windows ({count/len(y)*100:.1f}%)")
```

## 5. Machine Learning Classification

```python
# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

# Standardize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"Training set: {len(X_train)} samples")
print(f"Test set: {len(X_test)} samples")
```

```python
# Train multiple classifiers
classifiers = {
    'SVM (RBF)': SVC(kernel='rbf', C=10, gamma='scale', random_state=42),
    'SVM (Linear)': SVC(kernel='linear', C=1, random_state=42),
    'Random Forest': RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42)
}

results = {}

for name, clf in classifiers.items():
    print(f"\n{'='*60}")
    print(f"Training {name}...")
    
    # Train
    clf.fit(X_train_scaled, y_train)
    
    # Predict
    y_pred = clf.predict(X_test_scaled)
    
    # Evaluate
    acc = accuracy_score(y_test, y_pred)
    results[name] = acc
    
    print(f"\nTest Accuracy: {acc*100:.2f}%")
    print(f"\nClassification Report:")
    print(classification_report(y_test, y_pred, 
                                target_names=['REST', 'CLICK', 'SQUEEZE'][:len(np.unique(y))]))

print(f"\n{'='*60}")
print("SUMMARY:")
for name, acc in results.items():
    print(f"{name:20s}: {acc*100:.2f}%")
```

```python
# Confusion matrix for best classifier
best_clf_name = max(results, key=results.get)
best_clf = classifiers[best_clf_name]
y_pred_best = best_clf.predict(X_test_scaled)

cm = confusion_matrix(y_test, y_pred_best)
cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Raw counts
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['REST', 'CLICK', 'SQUEEZE'][:len(np.unique(y))],
            yticklabels=['REST', 'CLICK', 'SQUEEZE'][:len(np.unique(y))],
            ax=axes[0], cbar_kws={'label': 'Count'})
axes[0].set_title(f'Confusion Matrix - {best_clf_name}\n(Raw Counts)', 
                  fontsize=12, fontweight='bold')
axes[0].set_ylabel('True Label', fontsize=11)
axes[0].set_xlabel('Predicted Label', fontsize=11)

# Normalized
sns.heatmap(cm_normalized, annot=True, fmt='.2%', cmap='Greens',
            xticklabels=['REST', 'CLICK', 'SQUEEZE'][:len(np.unique(y))],
            yticklabels=['REST', 'CLICK', 'SQUEEZE'][:len(np.unique(y))],
            ax=axes[1], cbar_kws={'label': 'Percentage'})
axes[1].set_title(f'Confusion Matrix - {best_clf_name}\n(Normalized)', 
                  fontsize=12, fontweight='bold')
axes[1].set_ylabel('True Label', fontsize=11)
axes[1].set_xlabel('Predicted Label', fontsize=11)

plt.tight_layout()
plt.show()
```

## 6. Deep Learning Classification (CNN)

Train a 1D CNN on raw signal windows for end-to-end learning

```python
# Optional: Install TensorFlow if not already installed
# !pip install tensorflow

try:
    import tensorflow as tf
    from tensorflow import keras
    from tensorflow.keras import layers
    
    print(f"TensorFlow version: {tf.__version__}")
    TF_AVAILABLE = True
except ImportError:
    print("TensorFlow not installed. Skip this section or install with: pip install tensorflow")
    TF_AVAILABLE = False
```

```python
if TF_AVAILABLE:
    # Prepare data for CNN (raw signal windows)
    X_cnn = []
    y_cnn = []
    
    for i in range(n_windows):
        start = i * window_size
        end = start + window_size
        
        segment = signal_centered[start:end]
        label = labels[start]
        
        X_cnn.append(segment)
        y_cnn.append(label)
    
    X_cnn = np.array(X_cnn)
    y_cnn = np.array(y_cnn)
    
    # Reshape for CNN (samples, timesteps, channels)
    X_cnn = X_cnn.reshape(X_cnn.shape[0], X_cnn.shape[1], 1)
    
    # Normalize
    X_cnn = (X_cnn - X_cnn.mean()) / X_cnn.std()
    
    # Train/test split
    X_train_cnn, X_test_cnn, y_train_cnn, y_test_cnn = train_test_split(
        X_cnn, y_cnn, test_size=0.3, random_state=42, stratify=y_cnn
    )
    
    print(f"CNN input shape: {X_train_cnn.shape}")
    print(f"Number of classes: {len(np.unique(y_cnn))}")
```

```python
if TF_AVAILABLE:
    # Build CNN model
    def build_eeg_cnn(input_shape, num_classes):
        model = keras.Sequential([
            # Conv block 1
            layers.Conv1D(32, kernel_size=5, activation='relu', input_shape=input_shape),
            layers.BatchNormalization(),
            layers.MaxPooling1D(pool_size=2),
            layers.Dropout(0.3),
            
            # Conv block 2
            layers.Conv1D(64, kernel_size=5, activation='relu'),
            layers.BatchNormalization(),
            layers.MaxPooling1D(pool_size=2),
            layers.Dropout(0.3),
            
            # Conv block 3
            layers.Conv1D(128, kernel_size=3, activation='relu'),
            layers.BatchNormalization(),
            layers.GlobalAveragePooling1D(),
            layers.Dropout(0.4),
            
            # Dense layers
            layers.Dense(64, activation='relu'),
            layers.Dropout(0.4),
            layers.Dense(num_classes, activation='softmax')
        ])
        
        model.compile(
            optimizer='adam',
            loss='sparse_categorical_crossentropy',
            metrics=['accuracy']
        )
        
        return model
    
    # Create model
    input_shape = (window_size, 1)
    num_classes = len(np.unique(y_cnn))
    
    model = build_eeg_cnn(input_shape, num_classes)
    model.summary()
```

```python
if TF_AVAILABLE:
    # Train model
    early_stop = keras.callbacks.EarlyStopping(
        monitor='val_loss', patience=15, restore_best_weights=True
    )
    
    history = model.fit(
        X_train_cnn, y_train_cnn,
        validation_split=0.2,
        epochs=100,
        batch_size=32,
        callbacks=[early_stop],
        verbose=1
    )
```

```python
if TF_AVAILABLE:
    # Evaluate CNN
    test_loss, test_acc = model.evaluate(X_test_cnn, y_test_cnn, verbose=0)
    print(f"\nCNN Test Accuracy: {test_acc*100:.2f}%")
    
    # Predictions
    y_pred_cnn = np.argmax(model.predict(X_test_cnn, verbose=0), axis=1)
    
    print(f"\nClassification Report (CNN):")
    print(classification_report(y_test_cnn, y_pred_cnn,
                                target_names=['REST', 'CLICK', 'SQUEEZE'][:num_classes]))
    
    # Plot training history
    fig, axes = plt.subplots(1, 2, figsize=(14, 4))
    
    axes[0].plot(history.history['accuracy'], label='Train')
    axes[0].plot(history.history['val_accuracy'], label='Validation')
    axes[0].set_xlabel('Epoch')
    axes[0].set_ylabel('Accuracy')
    axes[0].set_title('CNN Training Accuracy', fontweight='bold')
    axes[0].legend()
    axes[0].grid(True, alpha=0.3)
    
    axes[1].plot(history.history['loss'], label='Train')
    axes[1].plot(history.history['val_loss'], label='Validation')
    axes[1].set_xlabel('Epoch')
    axes[1].set_ylabel('Loss')
    axes[1].set_title('CNN Training Loss', fontweight='bold')
    axes[1].legend()
    axes[1].grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
```

## 7. Results Summary

```python
# Compare all models
all_results = results.copy()
if TF_AVAILABLE:
    all_results['CNN (1D)'] = test_acc

# Bar plot
fig, ax = plt.subplots(figsize=(12, 6))
models = list(all_results.keys())
accuracies = [all_results[m]*100 for m in models]

bars = ax.bar(models, accuracies, color='skyblue', edgecolor='navy', linewidth=1.5)

# Highlight best model
best_idx = accuracies.index(max(accuracies))
bars[best_idx].set_color('gold')
bars[best_idx].set_edgecolor('darkgoldenrod')

ax.set_ylabel('Accuracy (%)', fontsize=13, fontweight='bold')
ax.set_title('EEG BCI Classification Performance', fontsize=15, fontweight='bold')
ax.set_ylim(0, 105)
ax.axhline(y=50, color='red', linestyle='--', alpha=0.5, label='Chance (50%)')
ax.grid(True, alpha=0.3, axis='y')
ax.legend()

# Add value labels on bars
for i, (model, acc) in enumerate(zip(models, accuracies)):
    ax.text(i, acc + 2, f"{acc:.1f}%", ha='center', fontsize=11, fontweight='bold')

plt.xticks(rotation=15, ha='right')
plt.tight_layout()
plt.show()

print("\n" + "="*60)
print("FINAL RESULTS SUMMARY")
print("="*60)
for model, acc in all_results.items():
    marker = " ★" if acc == max(all_results.values()) else ""
    print(f"{model:20s}: {acc*100:6.2f}%{marker}")
print("="*60)
```

## 8. Save Model for Real-Time Use

```python
import pickle

# Save best traditional ML model
model_data = {
    'classifier': best_clf,
    'scaler': scaler,
    'model_name': best_clf_name,
    'accuracy': results[best_clf_name],
    'feature_names': [
        'mean', 'std', 'var', 'kurtosis', 'skew', 'max', 'min', 'ptp',
        'zcr', 'delta_power', 'theta_power', 'alpha_power', 'beta_power',
        'gamma_power', 'alpha_beta_ratio', 'theta_alpha_ratio', 'combined_ratio'
    ]
}

with open('eeg_bci_model.pkl', 'wb') as f:
    pickle.dump(model_data, f)

print("✓ Traditional ML model saved to: eeg_bci_model.pkl")

# Save CNN if available
if TF_AVAILABLE:
    model.save('eeg_bci_cnn.h5')
    print("✓ CNN model saved to: eeg_bci_cnn.h5")
```
